Mục tiêu bài toán: Xây dựng mô hình máy học để dự đoán khả năng xảy ra cơn đau tim (Heart Attack) dựa trên các thông số sức khỏe của bệnh nhân.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report


# doc file va loc du lieu trong
df = pd.read_csv('Heart Attack.csv')
display(df.head())
df = df.dropna() # Assign the result of dropna back to df

#chuyen du lieu chu sang so
le = LabelEncoder()
for col in df.columns :
  if df[col].dtype == 'object':
    df[col] = le.fit_transform(df[col])


# tach lay dac trung va nhan
X = df.drop(columns = 'class')
Y = df['class']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state= 42)

#chuan hoa du lieu
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# huan luyen mo hinh
model = SVC(kernel = 'rbf', C=1.0, random_state = 42)
model.fit(X_train,Y_train)

# kiem tra mo hinh
y_pred = model.predict(X_test)


print(accuracy_score(Y_test, y_pred))
print(classification_report(Y_test, y_pred))




```
